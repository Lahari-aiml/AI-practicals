```python
print("Welcome to my first AI Cleaning Robot")

room = ["D", "C", "D", "D", "C"]
print("Room before cleaning:", room)

def clean_spot(spot):
    if spot == "D":
        return "C"
    return spot

cleaned = 0

for i in range(len(room)):
    if room[i] == "D":
        room[i] = clean_spot(room[i])
        cleaned += 1

print("Room after cleaning:", room)
print("Total spots cleaned:", cleaned)

my_room = ["D", "D", "C", "D", "D", "D"]
print("\nMy room before cleaning:", my_room)

my_cleaned = 0

for i in range(len(my_room)):
    if my_room[i] == "D":
        my_room[i] = clean_spot(my_room[i])
        my_cleaned += 1

print("My room after cleaning:", my_room)
print("Total spots cleaned:", my_cleaned)

print("Congrats! You have made your first AI Cleaning Robot.")
```
