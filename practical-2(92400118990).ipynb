```python
print("hello! iam your treasure hunt robot")
print("lets find out the treasure hunt together")

map_rooms = {
    "A": ["B", "C"],
    "B": ["D"],
    "C": ["D"],
    "D": ["G"],
    "G": []
}

print("\nOur Map:")
print("        A(start)")
print("       / \\")
print("      B   C")
print("       \\ /")
print("        D")
print("        |")
print("        G (Treasure)")

print()
print("doors from each room:", map_rooms)


# BFS

def bfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop(0)

        if room in visited:
            continue

        visited.append(room)
        order.append(room)

        print("BFS robot checking:", room)

        if room == goal:
            print("BFS found the treasure!")
            return order

        for next_room in map_rooms[room]:
            if next_room not in visited:
                to_do.append(next_room)

    return order


bfs_order = bfs("A", "G")

print("\nBFS checked the rooms in this order:", bfs_order)
print("Number of rooms BFS checked:", len(bfs_order))


# DFS

def dfs(start, goal):
    to_do = [start]
    visited = []
    order = []

    while to_do:
        room = to_do.pop()

        if room in visited:
            continue

        visited.append(room)
        order.append(room)

        print("DFS robot checking:", room)

        if room == goal:
            print("DFS found the treasure!")
            return order

        for next_room in map_rooms[room]:
            if next_room not in visited:
                to_do.append(next_room)

    return order


dfs_order = dfs("A", "G")

print("\nDFS checked the rooms in this order:", dfs_order)
print("Number of rooms DFS checked:", len(dfs_order))


# Uniform Cost Search

door_cost = {
    ("A", "B"): 1,
    ("B", "D"): 1,
    ("A", "C"): 5,
    ("C", "D"): 1,
    ("D", "G"): 1
}


# Calculate path cost

def path_cost(path):
    total = 0

    for i in range(len(path) - 1):
        door = (path[i], path[i + 1])
        total += door_cost[door]

    return total


# Two possible paths

path1 = ["A", "B", "D", "G"]
path2 = ["A", "C", "D", "G"]

cost1 = path_cost(path1)
cost2 = path_cost(path2)


print("\nPath 1:", path1)
print("Cost of path 1:", cost1)

print("\nPath 2:", path2)
print("Cost of path 2:", cost2)

if cost1 < cost2:

    best_path = path1
    best_cost = cost1

    print("\nUCS selected Path 1")
    print("UCS path:", best_path)
    print("UCS total cost:", best_cost)

elif cost2 < cost1:

    best_path = path2
    best_cost = cost2

    print("\nUCS selected Path 2")
    print("UCS path:", best_path)
    print("UCS total cost:", best_cost)

else:

    best_path = path1
    best_cost = cost1

    print("\nBoth paths have the same cost")


# UCS selects the lowest-cost path

if cost1 < cost2:
    print("\nUCS selected Path 1")
    print("UCS path:", path1)
    print("UCS total cost:", cost1)

elif cost2 < cost1:
    print("\nUCS selected Path 2")
    print("UCS path:", path2)
    print("UCS total cost:", cost2)

else:
    print("\nBoth paths have the same cost")
```
